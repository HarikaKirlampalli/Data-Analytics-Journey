```python
import matplotlib.pyplot as plt
import numpy as np
products = ["Laptop","Mobile","Tablet","Headphones","Watch"]#Dataset
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
```

```python
# Line Chart - Used when we want to see how values change or show a trend
plt.plot(products, sales, marker="o")  # Plot the sales values to compare products and see the change
plt.title("Products vs Sales")  # Add a title to describe the chart
plt.xlabel("Products")  # X-axis shows the product names or categories
plt.ylabel("Sales")  # Y-axis shows the sales values
plt.show()  # Display the line chart
```

```python
# Double Line Chart - Used to compare two sets of values and see how both change
plt.plot(products, sales, marker="o", label="Sales")  # Plot sales values with circle markers to show each data point
plt.plot(products, profit, marker="*", label="Profit")  # Plot profit values with star markers to compare with sales
plt.title("Sales vs Profit")  # Add a title to show that sales and profit are being compared
plt.xlabel("Products")  # X-axis represents the products being compared
plt.ylabel("Amount")  # Y-axis represents the numerical sales and profit amounts
plt.legend()  # Display the labels of both lines so we know which line is Sales and which is Profit
plt.show()  # Display the double line chart
```

```python
#BarChart- Used to compare different categories or values easily
plt.bar(products,sales,label = "Sales")
plt.title("Product sales")
plt.xlabel("products")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# Clustered Bar Chart - Used to compare two or more values side by side for each category
x = np.arange(len(products))  # Create positions for each product on the X-axis
width = 0.35  # Set the width of each bar
plt.bar(x - width/2, sales, width, label="Sales")  # Place Sales bars slightly left of each product position
plt.bar(x + width/2, profit, width, label="Profit")  # Place Profit bars slightly right to compare with Sales
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
# Stacked Bar Chart - Used to show two or more values stacked together for each category
x = np.arange(len(products))  # Create positions for each product on the X-axis
width = 0.35  # Set the width of each bar
plt.bar(x, sales, width, label="Sales")  # Create the first bar showing sales values
plt.bar(x, profit, width, bottom=sales, label="Profit")  # Stack profit on top of sales for each product
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
#DAY-2
# Line Chart - Used to show how values change and to identify trends
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May","Jun"]
monthly_sales = [100,120,150,140,180,200]
plt.plot(months,monthly_sales,marker="*")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Area Chart is used to show changes in values over time with a filled area.
plt.plot(months,monthly_sales)
plt.fill_between(months,monthly_sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# Pie Chart - Used to show how much each category contributes to the total
import matplotlib.pyplot as plt  # Import Matplotlib for creating charts
import numpy as np  # Import NumPy for numerical operations
products = ["Laptop", "Mobile", "Tablet", "Headphones", "Watch"]  # Product names
sales = [120, 180, 90, 150, 110]  # Sales values of each product
profit = [30, 50, 20, 40, 25]  # Profit values of each product
plt.pie(               # Create a pie chart
    sales,            # Use sales values to create the pie slices
    labels=products,  # Display product names on the slices
    autopct="%1.1f%%"  # Display percentage for each product
)
plt.title("Product Sales Distribution")  # Add a title to the chart
plt.show()  # Display the pie chart
```

```python
# 1. Violin Plot - Used to show the distribution and spread of data
import pandas as pd
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Math": [65, 70, 75, 80, 85, 60, 90, 72, 78, 82],
    "Python": [70, 75, 80, 85, 60, 65, 69, 82, 75, 62],
    "Sql": [62, 95, 42, 58, 65, 83, 76, 90, 69, 84]
}
df = pd.DataFrame(data)
print(df)

```

```python
import matplotlib.pyplot as plt
plt.violinplot(df["Math"])
plt.title("Distribution of Math Marks")
plt.ylabel("Marks")
plt.show()
```

```python
# 2. Stem Plot - Used to show individual data values and their distribution
plt.stem(df["Student"], df["Math"])  # Create a stem plot using student names and Math marks
plt.title("Student Math Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#3.Contour Plot - Used to represent 3-dimensional data in a 2-dimensional graph
import numpy as np
import matplotlib.pyplot as plt
#x = np.arange(1, 11)
#y = np.arange(1, 11)
x=[1,2,3]
y=[10,20]
X, Y = np.meshgrid(x, y)
Z = X*2 + Y*2
plt.contour (X, Y, Z)
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# Heatmap - Used to represent data values using different colors
marks = df[["Math","Sql","Python"]].values
plt.imshow(marks,cmap="Greens") #cmap is used for colours
plt.title("Students Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("students")
plt.colorbar()
```

```python
angles = np.linspace(0, 2*np.pi, 6)
values = [10, 20, 15, 25, 18, 10]
plt.polar(angles, values)
plt.title("Polar Plot")
plt.show()
```

```python
#DAY-4
# Subplots - Used to display multiple charts in one figure
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]
fig, ax = plt.subplots(1, 2)
ax[0].plot(x, y)
ax[0].set_title("Line Chart")
ax[1].bar(x, y)
ax[1].set_title("Bar Chart")
plt.show()

```

```python
# 3D linePlot - Used to visualize data using X, Y and Z axes
import matplotlib.pyplot as plt
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
z = [5, 10, 15, 20, 25]
ax.plot(x, y, z)
plt.show()
```

```python
#3D GRID
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")# Create 3D axis
x = np.linspace(-5, 5, 20)# Create X and Y values
y = np.linspace(-5, 5, 20)
X, Y = np.meshgrid(x, y)# Create a grid using X and Y
Z = X**2 + Y**2
ax.plot_surface(X, Y, Z)
plt.show()
```

```python
# 3D Scatter Plot
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")# Create 3D axis
x = np.random.rand(30)
y = np.random.rand(30)
z = np.random.rand(30)
ax.scatter(x, y, z)# Create 3D scatter plot
ax.set_title("3D Scatter Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
plt.show()
```

```python
# 3D Bar Plot
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
x = np.array([1, 2, 3, 4])# X, Y positions
y = np.array([1, 2, 3, 4])
z = np.zeros(4)
dx = np.ones(4)# Bar width and height
dy = np.ones(4)
dz = np.array([10, 20, 15, 25])
ax.bar3d(x, y, z, dx, dy, dz)
ax.set_title("3D Bar Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Values")
plt.show()
```

```python
# 3D Surface Plot
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
# Create surface plot
ax.plot_surface(X, Y, Z)
ax.set_title("3D Surface Plot")
plt.show()
```
