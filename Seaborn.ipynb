```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Sales": [100, 120, 150, 180, 200],
    "Sales-2": [110, 130, 140, 190, 220]
}
sns.scatterplot(data=data, x="Sales", y="Sales-2")
plt.title("Sales vs Sales-2")
plt.xlabel("Sales")
plt.ylabel("Sales-2")
plt.show()
```

```python
data = sns.load_dataset("tips")
sns.scatterplot(data=data, x="total_bill", y="tip")
plt.title("Scatters")
plt.show()
print(data)
```

```python
#Barplot
data = sns.load_dataset("tips")
sns.barplot(data=data, x="day", y="total_bill")
plt.title("Bar Plot")
plt.show()
print(data)
```

```python
#lineplot
data = sns.load_dataset("tips")
sns.lineplot(data=data, x="total_bill", y="tip",hue="sex",marker=".",estimator="mean",errorbar=None)
plt.title("Line Plot")
plt.show()
print(data)
```

```python
#BoxPlot
data = sns.load_dataset("tips")
sns.boxplot(data=data, x="day", y="total_bill")
plt.title("Box Plot")
plt.show()
print(data)
```

```python
#violinPlot
data = sns.load_dataset("tips")  # Load the tips dataset
sns.violinplot(data=data, x="day", y="total_bill")  # Create violin plot
plt.title("Violin Plot")
plt.show()
print(data)
```

```python
# Histplot - Distribution of Total Bill
data = sns.load_dataset("tips")  # Load the tips dataset
sns.histplot(data=data, x="total_bill")  # Create histogram for total bill
plt.title("Distribution of Total Bill")  # Add title
plt.show()
print(data)
```

```python
# Scatter Plot
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
sns.scatterplot(data=df, x="median_income", y="median_house_value")
plt.title("Scatter Plot")
plt.show()
print(df)
```

```python
# Box Plot
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
sns.boxplot(data=df, y="median_house_value")  # Create box plot
plt.title("Box Plot")
plt.show()
print(df)
```

```python
#Barplot
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
sns.barplot(data=df, x="housing_median_age", y="median_house_value")  # Create bar plot
plt.title("Housing Age vs House Value")
plt.show()
print(df)
```

```python
#Lineplot
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
sns.lineplot(data=df, x="housing_median_age", y="median_house_value")  # Create line plot
plt.title("Housing Age vs House Value")  # Add title
plt.show()
print(df)
```

```python
#Histplot
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
sns.histplot(data=df, x="median_house_value")
plt.title("Distribution of House Value")
plt.show()
print(df)
```

```python
#violinplot
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
sns.violinplot(data=df, y="median_house_value")
plt.title("Distribution of House Value")
plt.show()
print(df)
```
