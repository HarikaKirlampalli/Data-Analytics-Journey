```python
#DAY-5
# Aggrigations and Statistical Operations(sum,min,max,mean,median,std,var,argmax,argmin)
import numpy as np
arr = np.array([10,20,30,40,50])#1D
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.var(arr))
print(np.percentile(arr,50))
```

```python
#Percentile & Quantile
arr = np.array([10,20,30,40,50])
print(np.percentile(arr,100))
np.quantile(arr,1) #Quantile is opposite to the percentile(percentile btw 0 to 100 amd quantile btw proportion 0 and 1)
```

```python
#Argmax - means it returns the position of the highest value
#Argmin
arr = np.array([10,30,35,90,5,40])#indexing 0 1 2 3
print(np.argmax(arr))
print(np.argmin(arr))
```

```python
#Log10
arr = np.array([10,100,1000])
print(np.log10(arr))
#Log2
print(np.log(arr))
print(np.log(2))
print("-----------------------")
#Round()
arr = np.array([10.23,52.6,45.52,30.31])
print(np.round(arr))
arr = np.array([10.235,52.628,45.547])
print(np.round(arr,2))
print("-----------------------")
#Floor() - for lower limit
arr = np.array([10.2,108,209,32.9])
print(np.floor(arr))
print("-----------------------")
#Ceil() - for upper limit
arr = np.array([10.2,108,209,32.9])
print(np.ceil(arr))
print("-----------------------")
#abs() - it is used to convert negtive to positive number
arr = np.array([2,-5,3,9,-8])
print(np.abs(arr))
print("-------------------------")
#Sort()
arr = np.array([15,25,75,10,35])
result = np.sort(arr) #Asec order
print(result)
print("--------------")
arr = np.array([15,25,75,10,35])
result = np.sort(arr)[::-1] #Dsec order
print(result)
```

```python
#DAY-6
#Concatenate and Stack
#For 1D
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])
result = np.concatenate((arr1,arr2))
print(result)
print(result.ndim)
```

```python
#Concatenate - np.concatenate and Stack - np.stack
#For 2D
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])
result = np.stack((arr1,arr2))
print(result)
print(result.ndim)
```

```python
#Horizontal concatenate(axis=1) - Means it adds the 2nd array beside the 1st array
import numpy as np
arr1 = np.array([
    [10,20],
    [30,40]])
arr2 = np.array([
    [50,60],
    [70,80]
])
result = np.concatenate((arr1, arr2), axis=1)
print(result)
print(result.ndim)
```

```python
#Vertical concatenate(axis=0) - Means it adds the 2nd array down to the 1st array
import numpy as np
arr1 = np.array([
    [10,20],
    [30,40]])
arr2 = np.array([
    [50,60],
    [70,80]
])
result = np.concatenate((arr1, arr2), axis=0)
print(result)
print(result.ndim)
```

```python
# Stack - np.stack : it changes dimensions
import numpy as np
arr1 = np.array([10,20,30]) #Here it is changing 1D to 2D
arr2 = np.array([40,50,60])
result = np.stack((arr1,arr2),axis=0) #for rows
print(result)
print(result.ndim)
```

```python
# 1D to 2D - for columns
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])
result = np.stack((arr1,arr2),axis=1)
print(result)
print(result.ndim)
```

```python
#2D to 3D -For rows
import numpy as np
arr1 = np.array([
    [1,2,3],
    [4,5,6]
])
arr2 = np.array([
    [10,20,30],
    [40,50,60]
])
result = np.stack((arr1,arr2),axis=0)
print(result)
```

```python
#2D to 3D - For columns
import numpy as np
arr1 = np.array([
    [1,2,3],
    [4,5,6],
    [8,9,0]
])
arr2 = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
result = np.stack((arr1,arr2),axis=1) #Here all 1st rows will be one array and all 2nd rows will be another array
print(result)
```

```python
#VStack - Means vertical stack - dont need to specify axis
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])
result = np.vstack((arr1,arr2))
print(result)
print(result.ndim)#Here it changes 1D to 2D
```
