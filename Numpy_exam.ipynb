```python
# 18. Add 25 between 20 and 30, remove 40, and find the total, smallest value, largest value, and average of the resulting array.
arr = np.array([10, 20, 30, 40, 50])
arr = np.insert(arr, 2, 25)
print(arr)
arr = np.delete(arr, 3)
print(arr)
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
```

```python
# 19. Find the variance, standard deviation, 25th percentile, 75th percentile, position of the highest mark, position of the lowest mark, and calculate the base-10 logarithm of each value.
marks = np.array([45, 56, 67, 72, 81, 90, 95])
print(np.var(marks))
print(np.std(marks))
print(np.percentile(marks, 25))
print(np.percentile(marks, 75))
print(np.argmax(marks))
print(np.argmin(marks))
print(np.log10(marks))
```

```python
# 20. Combine A and B into one continuous array, arrange A and B one below the other,
# arrange A and B along a newly created dimension, combine all three arrays vertically,
# combine all three arrays horizontally, and divide the combined 1D array into 3 equal parts.
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
C = np.array([7, 8, 9])
print(np.concatenate((A, B)))
print("-------------------")
print(np.vstack((A, B)))
print("-------------------")
print(np.stack((A, B)))
print("-------------------")
print(np.vstack((A, B, C)))
print("-------------------")
print(np.hstack((A, B, C)))
print("-------------------")
arr = np.concatenate((A, B, C))
print(np.split(arr, 3))
```
